```python
from dateutil import parser
```

```python
# Define the function "calculate_days"
def calculate_days(dates_string):
    dates = dates_string[6:].split("-")
    dateOne = parser.parse(dates[0])
    dateTwo = parser.parse(dates[1])
    return str(dateOne-dateTwo)
```

```python
phrase = 'Aired: Jun 18, 2022 - Aug 28, 2022'
# test that your function works

calculate_days(phrase)
```
